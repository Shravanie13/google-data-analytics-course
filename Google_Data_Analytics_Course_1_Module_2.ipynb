## 1. Two important frameworks: data life cycle vs. data analysis process

They are related, but they are **not the same** and should not be used interchangeably.

### Data life cycle — how data is handled over time
1. **Plan:** Decide what data is needed, how it will be managed, who is responsible, and what outcome is desired.
2. **Capture:** Collect or bring data in from sources such as public datasets, company files, forms, or databases.
3. **Manage:** Store, maintain, protect, and organize data. Consider security, privacy, quality, and access.
4. **Analyze:** Use data to answer questions, solve problems, make decisions, and support goals.
5. **Archive:** Keep data that is no longer actively used but may be needed for future reference or legal/business reasons.
6. **Destroy:** Securely and appropriately delete data when it is no longer required, including copies where applicable.

### Example: an electricity provider
- Plan what information could explain household energy use: annual consumption, building type, and devices used.
- Capture the information from suitable sources.
- Manage it securely and maintain its quality.
- Analyze patterns to find ways customers could save energy.
- Archive older records that must be retained.
- Securely destroy records that no longer need to be kept.

### Data analysis process — how an analyst completes an analysis
1. **Ask:** Define the problem and understand stakeholder expectations.
2. **Prepare:** Identify, collect, and store the data needed.
3. **Process:** Clean and transform the data so it is accurate and usable.
4. **Analyze:** Organize and examine the data to find patterns and draw conclusions.
5. **Share:** Communicate findings, often using visualizations.
6. **Act:** Use the insights to make decisions or take action.

| Data life cycle | Data analysis process |
|---|---|
| Plan | Ask |
| Capture | Prepare |
| Manage | Process |
| Analyze | Analyze |
| Archive | Share (not a direct equivalent) |
| Destroy | Act (not a direct equivalent) |

**Remember:** The two frameworks overlap in places, but there is no exact one-to-one mapping. The data life cycle focuses on the life of data; the analysis process focuses on the work of answering a question.

## 2. Variations of the data life cycle

Organizations adapt life-cycle stages to their work, industry, goals, and responsibilities.

| Organization/context | Stages or activities described |
|---|---|
| U.S. Fish and Wildlife Service | Plan, Acquire, Maintain, Access, Evaluate, Archive |
| U.S. Geological Survey (USGS) | Plan, Acquire, Process, Analyze, Preserve, Publish/Share |
| USGS cross-cutting activities | Describe (metadata/documentation), manage quality, backup and secure throughout the stages |
| Financial institutions (Strategic Finance example) | Capture, Qualify, Transform, Utilize, Report, Archive, Purge |
| Harvard Business School | Generation, Collection, Processing, Storage, Management, Analysis, Visualization, Interpretation |

Wildlife and geological organizations may emphasize preservation and historical records. Financial institutions may explicitly include purging. A research/teaching context may emphasize visualization and interpretation. The names and number of stages can differ, but responsible data handling remains important.

## 3. The six phases of the data analysis process in detail

### 3.1 Ask
- Define the problem clearly before jumping into data.
- Understand what stakeholders expect and what they need to know.
- Identify the current state, the desired/ideal state, and the gap or obstacle between them.
- Clarify the scope: what is included and what is outside the analysis.
- Turn the problem into focused questions.
- The **Five Whys** technique can help move from a symptom to a root cause by repeatedly asking why.

**Example — ticket queues:** A sports arena wants to reduce the time fans wait in ticket lines. The analyst clarifies the goal with the arena manager, executive sponsor, and sales partners, then determines which part of the ticketing experience is in scope.

**Example — business risk:** Confirm whether the analysis covers every kind of risk or only weather-related risks such as hurricanes and tornadoes.

### 3.2 Prepare
- Identify what data can answer the questions.
- Find and collect relevant data, and decide where it will be stored.
- Understand the types of data available.
- Consider whether to collect new data or responsibly use existing data, consulting data owners where appropriate.
- Aim for objective, unbiased data so decisions are fair and fact-based.

Useful data distinctions:
- **Quantitative data:** numerical information that can be counted or measured, such as a rating or number of responses.
- **Qualitative data:** descriptive information, such as written comments about an employee's experience.
- **Cross-sectional data:** information collected at one point in time or during one particular period.
- **Longitudinal data:** information collected over a longer period, allowing changes to be observed.

### 3.3 Process
- Clean and prepare data before analysis.
- Correct errors, typos, inconsistent formats, missing or inaccurate values, and duplicates where appropriate.
- Combine datasets when needed and transform data into a useful format.
- Investigate outliers—values that are unusually far from the rest and may affect results. Do not automatically delete them; understand them first.
- Check whether expected data is present, whether missingness is random or systematic, and whether values are coded correctly.
- Verify that the data is complete and correct, and communicate important cleaning decisions to stakeholders.

Cleaning is an analyst's first close interaction with a dataset. Understanding its structure, quirks, and limitations helps prevent misleading conclusions.

### 3.4 Analyze
- Organize and examine processed data to identify patterns, relationships, and trends.
- Use analysis to answer the original questions and support conclusions, predictions, or decisions.
- Common tools include spreadsheets and SQL; Python is introduced later in the program.
- Stay objective. Do not force the data to match what you expected to find.

A useful idea from the employee-survey lesson: analysts help tell the story in the data, but it is not the analyst's personal story. The analyst should amplify what the evidence shows.

### 3.5 Share
- Explain findings to stakeholders in a clear and useful way.
- Use visualizations—such as charts, graphs, maps, and tables—to make complex information easier to understand.
- Choose the right level of detail for the audience.
- Present the results clearly and be ready to answer questions.
- In the employee-survey example, leaders receive a high-level organization-wide picture before examining individual teams, helping avoid unexpected findings without context.

### 3.6 Act
- Apply insights to the original problem.
- Decide what changes, interventions, or next steps should happen.
- For an employee-engagement survey, leaders and teams use findings to choose organization-wide and team-level improvements.
- For a learner, this course phase connects to preparing for a job search and completing a case study for a portfolio.

**Important:** Collecting survey responses is not the finish line. The real value comes when the findings lead to meaningful action.

## 4. Applying the process: employee engagement survey

| Phase | What the analyst does in this example |
|---|---|
| Ask | Clarify the purpose of the survey, the problem to solve, what leaders hope to learn, and the scope. |
| Prepare | Decide which quantitative and qualitative information is needed; choose cross-sectional or longitudinal data; identify new or existing sources. |
| Process | Check expected fields, missing values, coding, unusual values, and data quality. Understand the dataset before drawing conclusions. |
| Analyze | Conduct analyses linked to the original questions and remain objective rather than seeking confirmation of assumptions. |
| Share | Present high-level findings to executives first, then provide appropriate team-level detail. |

Skipping phases can result in weak or misleading insights. A careful process takes time, but it makes the outcome more useful.

## 5. Common data analyst tools

### 5.1 Spreadsheets
A **spreadsheet** is a digital worksheet made up of rows, columns, and cells. Analysts use spreadsheets to collect, store, organize, sort, clean, and analyze data. They can also create charts and graphs.

Popular tools: **Microsoft Excel** and **Google Sheets**.

- **Formula:** a set of instructions used to perform a calculation using spreadsheet data. For example, `=A1+B1` adds two cell values.
- **Function:** a preset command that performs a particular task. Examples include `SUM`, `AVERAGE`, and `COUNT`.

### 5.2 Databases and query languages
A **database** is a collection of data stored in a computer system. Databases can store large amounts of structured information and organize it using rules and relationships.

A **query** is a request for data or information from a database. A **query language** is a programming language used to communicate with a database.

**SQL** means **Structured Query Language**. It is widely used to retrieve and manipulate data in relational databases. Common SQL operations introduced in the lesson include:
- `SELECT` — retrieve data.
- `INSERT` — add records.
- `UPDATE` — change existing records.
- `DELETE` — remove records.

Examples of database/query platforms mentioned: **MySQL**, **Microsoft SQL Server**, and **BigQuery**.

### 5.3 Data visualization tools
Data visualization means representing information graphically, for example with charts, graphs, maps, and tables. Visuals can make complex information easier to understand and help stakeholders draw conclusions and make informed decisions.

- **Tableau:** supports interactive visualizations and dashboards, including a drag-and-drop interface.
- **Looker:** connects to data in a database and supports visual exploration of query results, with access to underlying data.

### 5.4 Programming languages
**Python** and **R** can help analysts prepare, process, analyze, and share data. Python is also used in later parts of this certificate program.

### How the tools work together
An analyst might store a large dataset in a database, use SQL to retrieve the required subset, export it to a spreadsheet for additional inspection, and then create a dashboard in Tableau or Looker. Python or R may be used for additional processing or analysis.

## 6. Choosing the right tool for the job

| Feature | Spreadsheet | Database |
|---|---|---|
| Access | Through a spreadsheet software application | Usually accessed and managed through database software and query languages |
| Structure | Rows, columns, and cells | Structured data organized using rules and relationships |
| Organization | Cells in worksheets | Tables and other related structures/collections |
| Data size | Often suitable for small-to-medium datasets | Designed to manage large and complex datasets |
| Data entry | Often allows manual entry and flexible editing | Usually uses stricter rules and consistent data entry controls |
| Users | Often used by one person at a time in a simple workflow | Supports multiple users and controlled access |
| Control | Users directly edit the worksheet | A database management system controls storage, access, and operations |

**When to use a spreadsheet:** organizing, sorting, cleaning, and analyzing small-to-medium datasets; quick calculations; simple charts.

**When to use a database:** storing, managing, and querying large or complex datasets; supporting consistent data rules and multiple users.

You do not need to choose only one. They complement each other. A database can store the main data, and an analyst can export a relevant subset to a spreadsheet. Conversely, spreadsheet data that becomes too large or complex can be imported into a database.

The best tool depends on the task, data size and complexity, collaboration needs, and the phase of the analysis process.

## 7. Data quality, privacy, and responsible handling

- **Data integrity:** keeping data accurate, consistent, and reliable throughout its use.
- **Credibility:** considering whether data and its source are trustworthy and suitable for the question.
- **Privacy:** protecting personal or confidential information and using it appropriately.
- **Security:** protecting data from unauthorized access, loss, or misuse.
- **Metadata:** information that describes a dataset, such as its source, meaning, format, or collection date.
- **Backup:** maintaining an additional copy so data can be recovered if needed.

During the life cycle, analysts should collect and use data responsibly, maintain quality, protect it, archive what must be retained, and securely destroy what is no longer needed. Paper records may need shredding; digital storage may require secure erasure methods.

## 12. Practice questions with answers

### Multiple-choice and short-answer practice

**1. Which sequence represents the data life cycle?**

A. Ask, Prepare, Process, Analyze, Share, Act  
B. Plan, Capture, Manage, Analyze, Archive, Destroy  
C. Plan, Ask, Capture, Share, Act, Destroy  
**Answer: B.** The data life cycle describes how data is handled over time.

**2. Which sequence represents the data analysis process?**

A. Ask, Prepare, Process, Analyze, Share, Act  
B. Plan, Capture, Manage, Analyze, Archive, Destroy  
**Answer: A.** It describes the steps analysts follow to answer a question and use the results.

**3. What is the main purpose of the Ask phase?**

**Answer:** Define the problem, understand stakeholder expectations, clarify scope, and identify the questions the analysis should answer.

**4. What is an outlier? Should it always be deleted?**

**Answer:** An outlier is a value unusually far from other values. No. First investigate whether it is an error or a genuine unusual observation; then decide how to handle it.

**5. What is the difference between a formula and a function?**

**Answer:** A formula is an instruction or expression for a calculation. A function is a preset command that performs a particular task. For example, `=A1+B1` is a formula, while `=SUM(A1:A5)` uses the SUM function.

**6. What does SQL stand for?**

**Answer:** Structured Query Language. It is used to communicate with databases.

**7. What is a query?**

**Answer:** A request for data or information from a database.

**8. Which tool is generally more suitable for a very large, complex dataset used by multiple people?**

**Answer:** A database, because it is designed for large structured datasets, controlled access, and multi-user workflows.

**9. Name two spreadsheet applications.**

**Answer:** Microsoft Excel and Google Sheets.

**10. Name two visualization tools from this module.**

**Answer:** Tableau and Looker.

**11. What is the difference between quantitative and qualitative data?**

**Answer:** Quantitative data is numerical and can be counted or measured. Qualitative data is descriptive, such as written feedback.

**12. What is the difference between cross-sectional and longitudinal data?**

**Answer:** Cross-sectional data is collected at one point or period; longitudinal data is collected over time.

**13. Why should analysts communicate with stakeholders during data cleaning?**

**Answer:** To explain important changes and assumptions, verify that the data is suitable, and make the process transparent.

**14. Why is the Share phase important?**

**Answer:** Findings have value only when people can understand them. Clear communication and visuals help stakeholders make informed decisions.

**15. What happens in the Act phase?**

**Answer:** Stakeholders use the insights to address the original problem, make decisions, or introduce improvements.

**16. Can spreadsheets and databases be used together?**

**Answer:** Yes. Data can be stored in a database, queried, and exported to a spreadsheet for further analysis. Spreadsheet data can also be moved into a database when it grows too large or complex.

**17. Why should an analyst avoid trying to prove a personal assumption?**

**Answer:** Analysis should be objective. The conclusions should reflect what the evidence shows, not what the analyst hoped or expected to find.

**18. What is the purpose of archiving data?**

**Answer:** To retain data that is no longer actively used but may be needed for future reference or other requirements.

**19. What is the purpose of destroying data?**

**Answer:** To securely remove data that is no longer required and reduce privacy and security risks.

**20. Give one example of a programming language used in data analytics.**

**Answer:** Python or R.

## 13. Mini case study: improving employee engagement

Imagine a company wants to understand why employee engagement is falling.

1. **Ask:** Meet leaders and clarify what “engagement” means, what decision they need to make, and which employees or teams are in scope.
2. **Prepare:** Select relevant survey questions. Include measurable ratings and optional written comments. Decide whether one survey or repeated surveys over time are needed.
3. **Process:** Check missing responses, inconsistent labels, duplicate entries, unusual ratings, and whether the data was collected consistently.
4. **Analyze:** Compare results across relevant groups and time periods, while protecting privacy and avoiding unsupported conclusions.
5. **Share:** Present the overall picture with clear charts and explain limitations. Then provide suitable team-level findings.
6. **Act:** Work with leaders and teams to choose realistic improvements, assign responsibility, and later check whether the changes helped.

**Reflection:** If the company only collects survey responses but never acts on the results, it has not gained the full value of the analysis process.
