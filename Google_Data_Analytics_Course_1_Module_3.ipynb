## 1. The data analyst's toolbox

A data analyst uses different tools at different stages of analysis.

| Tool | What it helps with |
|---|---|
| Spreadsheets | Store, organize, sort, calculate, and visualize relatively small datasets |
| SQL | Retrieve and filter data from databases, especially large datasets |
| Data visualization tools | Present patterns, trends, comparisons, and findings clearly |

**Simple analogy:** A spreadsheet is like an organized notebook, SQL is like asking a librarian to find exactly the records you need, and a visualization is like turning the findings into a picture that tells a story.

The best tool depends on the size of the data, the task, the analysis process, and the audience.


## 2. Spreadsheet basics

A spreadsheet is an electronic document used to store, organize, calculate, and analyze data in a grid.

### Main parts
- **Cell:** One box in the spreadsheet where data is entered.
- **Column:** A vertical group of cells, identified by letters (A, B, C…).
- **Row:** A horizontal group of cells, identified by numbers (1, 2, 3…).
- **Cell reference:** The column letter followed by the row number. Example: `D3` means column D, row 3.
- **Attribute:** A characteristic or quality used to describe data; usually represented by a column.
- **Header / column name / label:** The heading at the top of a column that describes the data in it.
- **Observation:** All the attributes about one entity or item represented in a row. For example, one person's name, age, and favorite color in a single row form one observation.

### Example table

| First Name | Last Name | Siblings | Favorite Color | Favorite Dessert |
|---|---|---:|---|---|
| Willa | Stein | 2 | Blue | Ice cream |
| Tony | Chen | 1 | Green | Cake |
| Maya | Patel | 3 | Purple | Brownie |

In this example:
- `First Name` is an attribute/header.
- `A2` is a cell reference if the first name Willa is in cell A2.
- The complete row containing Willa is one observation.

### Entering and formatting data
1. Start a blank spreadsheet.
2. Enter column headings in the first row, such as `First Name` and `Last Name`.
3. Enter each person's information in the rows below.
4. Bold the headings so they are easy to distinguish.
5. Resize columns by dragging the right edge of a column heading.
6. Use text wrapping when a heading or value is too long.

### Text display options
- **Overflow:** Text may extend into an adjacent empty cell.
- **Wrap:** Text continues on another line within the same cell, increasing row height.
- **Clip:** Text is visually cut off to fit the cell width.

### Organizing data by sorting
Sorting arranges rows based on a selected column, such as names from A to Z or sibling counts from lowest to highest.

Example workflow:
1. Select the complete data range across all related columns. This keeps each person's information together.
2. Choose **Data → Sort range** (menu wording may differ by spreadsheet program).
3. Open advanced sorting options if available.
4. Check **Data has header row** so the heading is not treated as an ordinary record.
5. Choose the column to sort by, such as `Siblings`.
6. Choose A to Z (ascending) or Z to A (descending), then sort.

**Important:** Select the whole table, not just one column. Sorting only one column can separate values from the records they belong to.

### Selecting cells
- Click and drag to select a range.
- Use Shift with the mouse or arrow keys to extend a selection.
- Use Ctrl (Windows) or Command (Mac) to select separate, nonadjacent cells or ranges.

### Formulas
A **formula** is a set of instructions that performs a calculation or action using spreadsheet data. Formulas begin with `=` and can use cell references.

Example:
```text
=C2+C3+C4
```
This adds the values in cells C2, C3, and C4. If those cells contain sibling counts, the formula calculates the total number of siblings.

**Remember:** Google Sheets saves automatically to Google Drive. Other spreadsheet programs, such as Excel, generally require you to save the file.


## 3. Creating a chart from spreadsheet data

A spreadsheet can turn table data into a chart.

Basic process:
1. Select the relevant data, including category labels and values.
2. Use the spreadsheet's **Insert → Chart** option (menu wording can vary).
3. Review the suggested chart.
4. Use the chart editor to change the chart type or adjust its appearance.
5. Add or improve titles and labels so the chart is clear.
6. Keep the final chart neat and professional.

### Bar chart and column chart
- A **column chart** displays bars vertically.
- A **bar chart** displays bars horizontally.
- Both can compare values across categories, such as the ages of different people.

Different spreadsheet programs may have different menus, but many include built-in charting and editing features.


## 4. Introduction to SQL

**SQL** stands for **Structured Query Language**. It is used to communicate with databases and retrieve, filter, and work with data stored in tables. Some people pronounce SQL as “sequel.”

### Why use SQL?
Spreadsheets are useful for smaller datasets, but very large datasets can become difficult or slow to work with in a spreadsheet. SQL can search and retrieve specific information from large databases efficiently.

A database system must understand SQL. Many database systems use SQL, and the basic structure of queries is broadly similar across them.

### Key terms
- **Database:** An organized collection of data stored in a computer system.
- **Table:** Data arranged in rows and columns within a database.
- **Query:** A request for data or information from a database.
- **Syntax:** The predetermined structure and rules of a language, including required words, symbols, punctuation, and their placement.
- **String:** Text data, such as a person's name.

### The three basic SQL building blocks
| SQL keyword | Purpose | Question it answers |
|---|---|---|
| `SELECT` | Chooses the columns to return | What information do I need? |
| `FROM` | Identifies the table | Where is the information stored? |
| `WHERE` | Filters rows using conditions | Which records should be returned? |

### Basic query structure
```sql
SELECT column_name
FROM table_name
WHERE condition;
```

The query is commonly written on separate lines to make it easier to read and troubleshoot.


## 5. SQL examples: SELECT, FROM, and WHERE

### Example 1: Return every column
```sql
SELECT *
FROM movies;
```
The asterisk (`*`) means all columns. This asks for every column from the `movies` table.

### Example 2: Filter for action movies
```sql
SELECT *
FROM movies
WHERE genre = 'Action';
```
- `SELECT *` returns all columns.
- `FROM movies` identifies the table.
- `WHERE genre = 'Action'` returns only rows whose genre is Action.

### Example 3: Select one column
```sql
SELECT first_name
FROM customer_data.customer_name
WHERE first_name = 'Tony';
```
This selects only the `first_name` column from the `customer_name` table in the `customer_data` dataset, and returns rows where the first name is Tony.

The expression `customer_data.customer_name` uses the dataset name, a dot, and the table name.

### Example 4: Select multiple columns
Separate column names with commas:
```sql
SELECT
    customer_id,
    first_name,
    last_name
FROM
    customer_data.customer_name
WHERE
    first_name = 'Tony';
```

Choose only the columns needed for the analysis. Extra columns can make results harder to work with and may use more resources.

### Example 5: Multiple conditions with AND
```sql
SELECT
    customer_id,
    first_name,
    last_name
FROM
    customer_data.customer_name
WHERE
    customer_id > 0
    AND first_name = 'Tony'
    AND last_name = 'Magnolia';
```
All conditions joined by `AND` must be true for a row to be returned.

### Example 6: Filter employee records
```sql
SELECT *
FROM Employee
WHERE jobCode = 'SFI';
```
This returns employee records whose job code is SFI.

### Example 7: Exclude interns and find salaries at or below $30,000
```sql
SELECT *
FROM Employee
WHERE jobCode <> 'INT'
  AND salary <= 30000;
```
- `<>` means **not equal to**.
- `<=` means **less than or equal to**.
- `AND` means both conditions must be true.

In the course example, this could identify non-intern employees earning $30,000 or less. Such information might help an organization investigate possible salary inequities and discuss whether salary adjustments are appropriate.


## 6. More SQL features

### Formatting and capitalization
SQL keywords are often written in uppercase (`SELECT`, `FROM`, `WHERE`) to make them stand out. Many SQL systems do not require uppercase keywords, and extra spaces or line breaks usually do not change the meaning of a query. Consistent formatting improves readability and makes errors easier to find.

### Semicolons
A semicolon (`;`) commonly marks the end of a SQL statement. It is recommended by the ANSI SQL-92 standard and is required by some systems or contexts, although some systems accept a query without one.

### Comparison and logical operators
| Operator | Meaning | Example |
|---|---|---|
| `=` | Equal to | `genre = 'Action'` |
| `<>` | Not equal to | `jobCode <> 'INT'` |
| `>` | Greater than | `customer_id > 0` |
| `<` | Less than | `salary < 30000` |
| `>=` | Greater than or equal to | `score >= 70` |
| `<=` | Less than or equal to | `salary <= 30000` |
| `AND` | Both conditions must be true | `salary <= 30000 AND jobCode <> 'INT'` |
| `OR` | At least one condition must be true | `city = 'Austin' OR city = 'Dallas'` |
| `NOT` | Negates a condition | `NOT (genre = 'Action')` |

### LIKE and the percent wildcard
`LIKE` is used for pattern matching in text.

```sql
SELECT *
FROM Employee
WHERE lastName LIKE 'Ch%';
```
This looks for last names beginning with `Ch`, such as Chavez or Chen.

The `%` wildcard represents zero or more characters in common SQL pattern matching. Wildcard behavior can vary in other contexts, so use the syntax supported by the database system.

### Comments
Comments explain the purpose of a query, table, column, condition, or step. They help the original writer and other analysts understand the code later.

Common forms:
```sql
-- This is a single-line comment
```
```sql
/* This is a block comment */
```

Example:
```sql
-- Retrieve basic customer information
SELECT
    customer_id, -- customer identifier
    first_name,  -- customer's first name
    last_name    -- customer's last name
FROM
    customer_data.customer_name;
```
Use comment syntax supported by the database. `--` is broadly supported; `/* ... */` is also common. `#` is not supported as a SQL comment marker by every database (for example, it is not standard SQL comment syntax).

### Aliases with AS
An **alias** is a temporary alternative name for a table or column in a query. `AS` introduces the alias.

```sql
SELECT
    actual_column_name AS clearer_column_name
FROM
    actual_table_name AS t;
```

An alias does not rename the actual database column or table. It applies to the query result or query reference and is temporary.


## 7. Data visualization: meaning and importance

**Data visualization** is the graphical representation of information and data. It turns numbers and data points into visuals such as charts, graphs, maps, and dashboards.

### Why visualization matters
- Makes information easier and quicker to understand.
- Helps people notice patterns, trends, and comparisons.
- Communicates findings to stakeholders who may not have time to inspect detailed tables.
- Makes analysis more engaging and memorable.
- Supports decisions with evidence rather than relying only on assumptions.

A table is useful for storing and inspecting detailed values. A visualization often makes patterns and trends easier to see at a glance.

### Stakeholders
**Stakeholders** are people or groups who have an interest in, are affected by, or make decisions based on the analysis. Examples include managers, executives, customers, and marketing teams.

A data analyst should design visuals for the audience's needs and make the audience's time worthwhile.

### Florence Nightingale: historical example
During the Crimean War in the 1850s, Florence Nightingale examined causes of soldiers' deaths. She found that many deaths were caused by preventable conditions. She created a chart showing deaths over time; the large blue sections represented preventable deaths in the course illustration. Her visualization helped communicate the problem to hospital administrators and contributed to changes in patient care.

**Lesson:** A clear visual can make evidence easier to understand and can help persuade decision-makers to act.


## 8. Planning a data visualization

A useful planning process has three main steps.

### Step 1: Explore the data for patterns
- Ask the manager or data owner for access to relevant data.
- Review the data and understand what it contains.
- Look for patterns, relationships, unusual values, and trends.
- In the clothing distributor example, the analyst examines sales records and website analytics, including customer visits, purchases, spending, and geography.

### Step 2: Plan your visuals
Think about the audience and what they need to learn. In the clothing distributor example, the audience is sales-oriented, so the visuals should:
- Show sales numbers over time.
- Connect sales to location.
- Show the relationship between sales and website use.
- Show which customers contribute to growth.

### Step 3: Create your visuals
- Choose visual formats that communicate the findings.
- Try different chart types.
- Adjust and refine the charts.
- Combine multiple visuals when one chart cannot tell the whole story.
- Use spreadsheet chart features or more specialized visualization tools.

**Remember:** Visualization is a process. The first chart is not always the best chart. Test, review, and improve it.

### SQL query checklist
Before running a query, ask:
1. Did I choose the correct columns after `SELECT`?
2. Did I specify the correct table after `FROM`?
3. Are my `WHERE` conditions correct?
4. Did I use `AND` / `OR` appropriately?
5. Are text values enclosed in single quotes?
6. Is the syntax supported by my database?
7. Is the query formatted clearly and terminated with a semicolon when appropriate?

### Visualization checklist
Before sharing a chart, ask:
1. Who is the audience?
2. What question should the visual answer?
3. Does the chart type fit the data and message?
4. Are the labels and title clear?
5. Can the audience understand the main point quickly?
6. Is the chart accurate and uncluttered?
7. Would an interactive dashboard help the audience explore details?
