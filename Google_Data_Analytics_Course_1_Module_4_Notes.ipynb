## 4. The Power of Data in Business

Businesses usually have something they need to understand or solve.

Three useful terms are:

### Issue

Example:
- "Why are weekend visitors changing?"

### Question
Something designed to **discover information**.

### Problem
An **obstacle or complication that needs to be worked out**.

Example:
- "The zoo does not know how many employees it needs on rainy days."

### Business task
The **question or problem that data analysis resolves for a business**.

This is a key Module 4 term.

### Example

Zoo problem:
> Staffing is unpredictable.

> Analyze historical weather and attendance data to identify patterns that can improve staffing decisions.


## 5. Data-Driven Decision-Making (DDDM)

**Data-driven decision-making** means using facts discovered through data analysis to guide business strategy.

### Simple flow

**Business problem → Data → Analysis → Insight → Decision → Action**

Example:

Rainy weather  
↓  
Historical attendance data  
↓  
Analyze relationship between rain and attendance  
↓  
Discover lower attendance on rainy days  
↓  
Plan fewer staff during predictable low-attendance periods  
↓  
Better staffing and visitor experience

### Why not rely only on intuition?

Observation and experience can be useful, but relying only on memory or gut feelings may give an incomplete picture.

Data can reveal:

- Patterns we did not notice.
- Relationships between factors.
- New explanations.
- Better solutions.

### Important

Data does not automatically make a decision. Analysts provide evidence and insights that help people make better-informed decisions.

## 7. Fairness in Data Analytics

### Definition

**Fairness** is a quality of data analysis that does not create or reinforce bias.

An analysis can contain **true data** and still produce an unfair conclusion.

Why?

Because the analysis may ignore important social, environmental, or contextual factors.

### Example: Company "boys' club"

Suppose data shows that men are the only employees succeeding.

A company might incorrectly conclude:

> "We should hire more men."

The data may be factually accurate, but the conclusion is unfair if the company ignores a toxic culture that creates barriers for other gender identities.

A fairer approach would investigate the workplace culture and barriers before deciding what action to take.

### Core principle

Ask:

- Who is represented?
- Who is missing?
- What factors affect the results?
- Could the data reflect an existing bias?
- Could our conclusion create or reinforce unfairness?


## 8. Consider Fairness: Five Best Practices

### 1. Consider all available data

Do not ignore relevant information just because it does not match expectations.

**Example:** When analyzing holiday traffic, include weather if weather affects traffic.

### 2. Identify surrounding factors

Look at the context around the data.

**Example:** When planning employee leave, consider cultural and religious holidays, not only national holidays.

### 3. Include self-reported data

**Self-reporting** is a data collection technique where participants provide information about themselves.

This can reduce observer bias.

**Example:** Ask customers about their demographics instead of having sales staff guess.

### 4. Use oversampling effectively

**Oversampling** means increasing the sample size of nondominant or underrepresented groups.

It can help represent these groups more fairly and address imbalanced datasets.

**Example:** If older adults are underrepresented in a fitness study, deliberately include more participants aged 70+.

### 5. Think about fairness from beginning to end

Fairness should be considered during:

**Collection → Cleaning → Processing → Analysis → Presentation → Action**

Do not wait until the final report to think about fairness.

## 10. Alex – Fair and Ethical Data Decisions

Alex is a research scientist at Google working with the Ethical AI team.

The team studies how AI affects society and whether it helps or harms marginalized communities.

### Data ethics

**Data ethics** means thinking about what is good and right when collecting, handling, and using data.

Important questions include:

- Who is collecting the data?
- Why are they collecting it?
- How is it being collected?
- What purpose will it serve?
- Will the people represented in the data benefit?

### Beneficence

**Beneficence** means going beyond simply avoiding harm and actively using data to improve people's lives.

### Privacy and protection

Personal information can include:

- Images
- Voices
- Text
- Other identifying information

This information should be protected and not casually shared.

### Consent and user control

People should have meaningful ways to:

- Give consent before data is collected.
- Understand/control how their data is used.
- Request that data be revoked or removed where appropriate.

### Central ethical lesson

**Behind every piece of personal data, there are real people.**

## 17. Simple Python Example

The following example demonstrates the idea of comparing sales on rainy and non-rainy days.

It is only a learning example; real analysis would require a larger and more carefully collected dataset.


````python
import pandas as pd

data = pd.DataFrame({
    "rain": ["Yes", "No", "Yes", "No", "No", "Yes"],
    "sales": [120, 250, 140, 300, 280, 110]
})

print(data.groupby("rain")["sales"].mean())
````

### What the code does

- Creates a small DataFrame.
- Groups records by rain status.
- Calculates average sales for each group.

The same logic can be performed with spreadsheet tools or SQL.

### SQL version

```sql
SELECT
    rain,
    AVG(sales) AS average_sales
FROM cafe_sales
GROUP BY rain;
```

### Spreadsheet idea

If `A2:A100` contains rain status and `B2:B100` contains sales, a spreadsheet can calculate an average for rainy days with:

```text
=AVERAGEIF(A2:A100,"Yes",B2:B100)
```
