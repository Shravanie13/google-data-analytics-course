## 1. Introduction to Data Analytics

### What is data?
**Data** is a collection of facts. It can include numbers, words, measurements, dates, observations, images, and other information.

Everyday examples:
- Daily expenses
- Sleep hours
- Exam marks
- Shopping history
- Number of steps walked

Data becomes useful when we examine it and understand what it tells us.

### What is data analysis?
**Data analysis** is the collection, transformation, and organization of data to draw conclusions, make predictions, and support informed decision-making.

In simple words: **data analysis means turning data into insights.**

**Example — sleep analysis:** Suppose you record your sleep hours and mood every day. After organizing the information, you notice that you tend to feel better after seven hours of sleep. You have collected data, organized it, identified a pattern, and drawn a conclusion.

### Why businesses use data analysis
- Improve processes
- Identify opportunities and trends
- Understand customers
- Develop products and services
- Improve customer experience
- Make thoughtful decisions and remain competitive

### What does a data analyst do?
A **data analyst** collects, transforms, and organizes data to draw conclusions, make predictions, and support informed decisions.

Typical work:
1. Understand the business problem.
2. Collect relevant data.
3. Clean and organize it.
4. Examine patterns and relationships.
5. Communicate findings.
6. Help decision-makers take action.

## 2. Four Common Types of Data

| Type | Meaning | Examples |
|---|---|---|
| **Numerical** | Numbers that can be measured, counted, or calculated | Age, height, price, quantity |
| **Text/String** | Words, sentences, or identifiers | Product name, address, phone number, order ID |
| **Date** | A point in time | Order date, shipping date, birth date |
| **Categorical** | Labels or groups | Colour, Gold/Silver/Bronze, Excellent/Good/Poor |

**Important:** A phone number contains digits, but it is treated as text because we do not add or average phone numbers.

A **dataset** is a collection of data that can be manipulated or analyzed as one unit.

## 3. The Six Phases of Data Analysis

**Ask → Prepare → Process → Analyze → Share → Act**

### 1. Ask — What is the problem?
Understand the business challenge, define the objective, ask relevant questions, collaborate with stakeholders, and decide what success looks like.

### 2. Prepare — What data do we need?
Identify data sources, collect relevant information, check usefulness and accuracy, plan storage, consider access rules, and set a timeline.

### 3. Process — Is the data ready?
Clean and organize data. Remove duplicates, correct errors, identify missing values, fix inconsistent formats, and remove irrelevant information.

### 4. Analyze — What does the data tell us?
Examine data to find patterns, relationships, and trends. Analysts may calculate averages, count categories, compare groups, and create visualizations.

### 5. Share — Who needs to know?
Communicate findings through reports, presentations, charts, and dashboards. Explain results with enough context for decision-makers.

### 6. Act — What should we do?
Use the findings to make improvements or solve the original problem.

| Phase | Main question | Main activity |
|---|---|---|
| Ask | What is the problem? | Define the challenge |
| Prepare | What data is needed? | Collect information |
| Process | Is the data ready? | Clean and organize |
| Analyze | What does it show? | Find insights |
| Share | Who needs to know? | Communicate results |
| Act | What should happen next? | Implement solutions |

**Remember:** The process is a framework, not always a perfectly straight line.

## 4. Case Study: People Analytics

**People analytics** (also called HR analytics or workforce analytics) means collecting and analyzing workforce data to improve employee experiences and company operations.

### Business problem
A company notices that many new employees leave before completing their first year. It asks: **How can we improve retention among new employees?**

- **Ask:** Discuss possible causes with managers. Ask about onboarding, hiring, compensation, employee needs, and desired retention improvements.
- **Prepare:** Create a project timeline and an employee survey. Decide what information to collect and who can access it.
- **Process:** Obtain consent, explain how information will be protected, restrict access, clean responses, and summarize data where needed to protect individual privacy.
- **Analyze:** Discover that long, complicated hiring processes are associated with employees more likely to leave, while efficient and transparent evaluations are associated with employees more likely to stay.
- **Share:** Present findings to appropriate managers with context so they can discuss improvements with their teams.
- **Act:** Improve hiring and evaluation practices, repeat the survey annually, and compare results. The case study reports improved retention after changes.

**Ethical responsibility:** Protect employee information, limit access, and report findings honestly—even when results are unexpected.

## 5. Case Study: Geo-Flow, Inc.

Geo-Flow is a fictional technology company with low participation in its retirement contribution program. Management wants to know whether employee education could improve participation.

- **Ask:** Are employees participating? Do they understand the program and employer matching contributions?
- **Prepare:** Collect employee demographics, salary levels, and current retirement contributions.
- **Process:** Remove duplicates and records of former employees; organize data by age, department, and length of employment.
- **Analyze:** Find that some groups contribute less and may not understand the employer matching contribution.
- **Share:** Present bar and pie charts to management.
- **Act:** Provide targeted education to groups with low contributions. Contributions later increase among those groups.

### Iteration
**Iteration** means returning to an earlier phase or repeating steps when necessary. For example, analysis may reveal that the wrong database was selected, or cleaning may show that the original question was unclear.

Do not skip steps just to finish quickly. Review work throughout the process.

## 6. Other Data Analysis Frameworks

Different organizations use different frameworks. Their number of steps varies, but they share common ideas: define a problem, prepare and process data, analyze it, communicate results, and use the findings.

| Framework | Steps | Main stages |
|---|---:|---|
| **Google** | 6 | Ask, Prepare, Process, Analyze, Share, Act |
| **EMC / Dell EMC** | 6 | Discovery, Pre-processing, Model Planning, Model Building, Communicate Results, Operationalize |
| **SAS** | 7 | Ask, Prepare, Explore, Model, Implement, Act, Evaluate |
| **Project-based process** | 5 | Identify problem, Design data requirements, Pre-process, Analyze, Visualize |
| **Big Data process** | 9 | Business case evaluation, Data identification, Acquisition/filtering, Extraction, Validation/cleaning, Aggregation/representation, Analysis, Visualization, Utilization |

### Important details
- **EMC:** Associated with David Dietrich; phases connect and repeat. Check that data is ready before modelling.
- **SAS:** Emphasizes repeatable, reliable results and evaluation. After acting, evaluate and return to an earlier step if needed.
- **Project-based:** Developed by Vignesh Prajapati; ends with visualization and does not list a separate Act phase.
- **Big Data:** Described by Thomas Erl, Wajid Khattak, and Paul Buhler in *Big Data Fundamentals: Concepts, Drivers & Techniques*. It divides data preparation into more detailed tasks.

The course reading also connects the history of organized data to ancient Egypt, where scribes documented calculations and information on papyri.

## 7. Data Ecosystem and Cloud

### Ecosystem
An **ecosystem** is a group of elements that interact with one another. Examples include a rainforest, the Australian outback, tadpoles in a puddle, or bacteria on skin.

### Data ecosystem
A **data ecosystem** is the collection of interacting elements used to produce, manage, store, organize, analyze, and share data.

It includes:
- Hardware
- Software
- People
- Storage systems
- Technology and analysis tools

### Cloud storage
Cloud storage keeps data online rather than only on a computer's physical hard drive. It can be accessed over the internet.

**Example:** A file saved only on a laptop is local. A file saved in Google Drive is stored in the cloud and can be accessed online from different devices.

### Examples across industries
- **Retail:** Customer details, purchases, and reviews can help predict demand and stock products.
- **Human resources:** Job postings, labour-market statistics, employment rates, and social media information can support recruiting and retention.
- **Agriculture:** Geological and weather patterns can help predict crop yields.
- **Environment:** Scripps Institution of Oceanography digitally monitors coral reefs, tracking organisms, growth, and changes in colonies.

## 8. Data Science vs. Data Analysis vs. Data Analytics

| Term | Meaning |
|---|---|
| **Data science** | Uses raw data to create new ways of modelling and understanding the unknown |
| **Data analysis** | Collects, transforms, and organizes data to draw conclusions, make predictions, and support decisions |
| **Data analytics** | The broader science of data, including the work, tools, and methods used to manage and understand data |

A data scientist may create new questions and models. A data analyst often answers existing questions using available data. Data analysis and the data ecosystem fit under the broader umbrella of data analytics.

### Decision Intelligence and specialization
Cassie, a Google Cloud Decision Intelligence lead, describes Decision Intelligence as combining applied data science with social and managerial sciences to turn data into impact.

She compares analysts to:
- **Explorers:** investigate unknown information.
- **Detectives:** examine evidence and clues.
- **Artists:** use creativity to discover and communicate insights.

No one person can know everything about the expanding data universe, so specialization matters.

| Field | Focus in Cassie's explanation |
|---|---|
| Statistics | Rigor: careful decisions under uncertainty |
| Machine learning / AI | Performance: automate many decisions under uncertainty |
| Analytics | Speed and exploration: discover insights when the questions or decisions are not fully known at the start |

Analytics may suit people who enjoy ambiguity, exploration, varied data sources, and open-ended questions. Perfectionism should not prevent exploration.

## 9. Data and Gut Instinct

**Gut instinct** is an intuitive feeling or understanding, often without a clear explanation.

Analysts and detectives are similar because both collect facts, examine evidence, ask questions, follow clues, and look for patterns.

### Why gut instinct alone can be risky
Relying only on experience can create bias, incorrect assumptions, and poor decisions.

**Restaurant example:** An entrepreneur and famous chef open a themed restaurant based on the chef's reputation and previous success. They do not properly research customers in the new location. The new customers have different preferences, fundraising fails, and the property is sold at a loss. Local research could have revealed the mismatch.

Experience can still help people notice patterns. The best approach is to use experience alongside evidence.

Questions to ask:
1. What results are needed?
2. Who will use or be informed by the results?
3. Am I answering the question being asked?
4. How quickly must the decision be made?

A rushed decision may rely more on experience; a project with sufficient time and resources can use a more data-driven approach.

## 10. Five Essential Analytical Skills

Analytical skills are qualities and characteristics associated with using facts to solve problems.

1. **Curiosity:** Wanting to learn, explore, investigate, and ask questions. Example: asking why sales dropped.
2. **Understanding context:** Understanding the conditions surrounding information and seeing the full picture. Example: considering holidays, audience demographics, and competing events when analyzing movie revenue.
3. **Technical mindset:** Breaking a task into smaller steps and working logically. Example: sorting bills by due date, adding amounts, checking the bank balance, and then paying.
4. **Data design:** How information is organized. Example: arranging phone contacts by first name, last name, or email so they are easy to find.
5. **Data strategy:** Managing the people, processes, and tools used in data analysis.

### Data strategy: three components
- **People:** Make sure people know how to use data and understand their responsibilities.
- **Processes:** Make the path to a solution clear and accessible.
- **Tools:** Choose technology suited to the task.

**Lawn-mowing example:** Read the mower manual (people/knowledge), wear protective gear and clear the lawn of sticks and rocks (process), and check the mower's fuel, oil, and condition (tools).

## 11. Five Aspects of Analytical Thinking

**Analytical thinking** means identifying and defining a problem, then solving it with data in an organized, step-by-step way.

1. **Visualization:** Graphical representation of information, such as charts, maps, and diagrams. It helps people understand and explain information.
2. **Strategy:** Decide what needs to be achieved and how. Stay focused and collect useful, high-quality data.
3. **Problem-orientation:** Identify, describe, and solve the problem. Keep the original problem in mind throughout the project.
4. **Correlation:** A relationship between two or more pieces of data. Example: rainy weather and umbrella sales.
5. **Big-picture and detail-oriented thinking:** Big-picture thinking sees the whole situation and opportunities; detail-oriented thinking examines the individual pieces needed to execute a plan.

**Important:** Correlation does not equal causation. Two things changing together does not prove that one caused the other.

## 12. Using Analytical Skills in a Business Scenario: Mega-Pik International

Mega-Pik is a fictional film company. Five of its last six releases barely broke even, and the sixth lost money. Competitors have used remakes of successful older films to attract new audiences. Mega-Pik hires an analytics company to use **exploratory data analysis (EDA)** to understand audience preferences.

### EDA objectives
- Identify factors linked to opening-weekend success.
- Understand the relationship between movie budget and revenue.
- Determine the most successful genres.

### Dataset fields
Movie name, release date, opening-night revenue, opening-weekend revenue, budget, marketing costs, ratings, and genre.

### Applying the five skills
- **Curiosity:** Ask whether budget relates to revenue, whether genre performance changes by season, and whether audience demographics are needed.
- **Context:** Consider release timing, holidays, competing events, demographics, and differences in vacation seasons across countries. Comparing multiple years can help avoid misleading conclusions.
- **Technical mindset:** Clean, organize, and prepare data systematically; choose tools to find and correct errors.
- **Data design:** Structure and sort the data to reveal patterns, such as revenue by genre.
- **Data strategy:** Choose people, processes, and tools based on the deliverable. A simple dashboard may suit Sheets or Excel; an automatically updating dashboard may need a tool such as Tableau. Prioritize work that supports the business goal.

## 13. The Five Whys: Root Cause Analysis

A **root cause** is the reason a problem occurs. Finding and addressing it can prevent the problem from happening again.

The **Five Whys** technique repeatedly asks “Why?” until the underlying cause is identified. It often takes five questions, but may take fewer or more.

### Case 1: Online grocery store
- Problem: Customers receive damaged products.
- Why 1? Products are not packaged properly.
- Why 2? Packers are inadequately trained.
- Why 3? Many packers are new and have not completed training.
- Why 4? New employees are already packing orders.
- Why 5? HR is redesigning training and supplied an insufficient one-page quick guide instead of proper training.

**Root cause:** The updated training was incomplete and new packers were not properly trained.

**Solution:** Complete the training program and train new packers.

### Case 2: Irrigation pump defects
- Problem: More pumps are defective.
- Why 1? Machines are not properly calibrated.
- Why 2? They were miscalibrated during maintenance.
- Why 3? The existing calibration method is unsuitable.
- Why 4? New software changed calibration requirements, but engineers did not know.
- Why 5? The installation team failed to share updated calibration procedures.

**Root cause:** Engineers lacked the correct calibration information after the software upgrade.

**Solution:** Give engineers the updated instructions.

### Benefits
- Finds the underlying cause.
- Helps prevent recurring problems.
- Encourages collaboration.
- Helps analysts understand other business areas.
- Works across industries.

## 14. Data-Driven Decision-Making

**Data-driven decision-making** means using facts to guide business strategy.

Data can help organizations:
- Gain insights.
- Check assumptions and theories.
- Understand opportunities and challenges.
- Support objectives and plans.
- Make decisions with greater confidence.
- Save time and effort.

### Dairy farmer example
Instead of guessing which ice-cream flavours customers want, a dairy farmer surveys people and uses their preferences to select flavours.

### HR benefits example
An HR director assumes casual dress is the most valued workplace perk. A structured survey reveals that employees value free public transportation cards more. The director had missed this because she drives to work.

**Lesson:** Personal assumptions may not represent what other people want. Evidence can reveal the difference.

## 15. Real-World Case Studies

### Google People Analytics: Do managers add value?
Google examined performance reviews and employee surveys to understand whether managers improve teams.

The first results were too general. The team divided results into **quartiles**, meaning four equal parts:
- Top quartile: teams with the best-rated managers.
- Bottom quartile: teams with the lowest-rated managers.

Teams with the best managers were significantly happier, more productive, and more likely to stay. This supported the conclusion that managers add value.

Google then:
- Created an awards program where employees nominated favourite managers and gave evidence.
- Interviewed managers in the top and bottom quartiles.
- Identified effective management behaviours and common areas for improvement.
- Shared findings and developed an evaluation procedure based on those qualities.

### Nonprofit journalism
Nonprofits work on social causes such as food security, education, and the arts. Journalists can raise awareness and encourage people to learn, volunteer, donate, or make changes.

Analysts tracked story topics, clicks, web traffic, comments, and shares. They used the findings to recommend ways journalists could increase the impact of their stories and help nonprofits motivate public action.

## 16. Gap Analysis

**Gap analysis** is a method for examining and evaluating the current state of a process to identify opportunities for improvement in the future.

It compares:
- **Current state:** Where are we now?
- **Desired future state:** Where do we want to be?
- **Gap:** What needs to change?

**Example:** A company currently takes five days to complete an order, but its target is two days. The three-day difference is the gap to investigate and improve.

**Assessment answer:** A method for examining and evaluating how a process works currently in order to get to an improved future state is **gap analysis**.

## 18. Assessment-Taking Strategies
